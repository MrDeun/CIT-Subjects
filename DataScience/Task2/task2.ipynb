```python
import pandas

excel_sheet = pandas.ExcelFile("samotnosc.xlsx")
# df = pandas.read_excel("samotnosc.xlsx",sheet_name=excel_sheet.sheet_names[1],header=1)
human_df = pandas.read_excel("samotnosc.xlsx",sheet_name=excel_sheet.sheet_names[0],header=1)
df = human_df.sort_values(by="waga",ascending=False)
# print(excel_sheet.sheet_names)
print(df.head(20))
```

```python
import matplotlib.pyplot as plt
import pandas

counts = pandas.crosstab(human_df["klasa"], human_df["agecat"])
counts.plot(kind="bar", figsize=(12, 6))
plt.xlabel("Miejsce zamieszkania")
plt.ylabel("Liczba osób")
plt.title("Miejsce zamieszkania według kategorii wieku")
plt.xticks(rotation=30, ha="right")
plt.legend(title="Kategoria wieku")
plt.tight_layout()
plt.show()
```

```python
questions = pandas.read_excel("samotnosc.xlsx", sheet_name=excel_sheet.sheet_names[0], header=None, nrows=2).T
questions.columns = ["Pytanie", "Kod"]
questions = questions[["Kod", "Pytanie"]].reset_index(drop=True)
for _, row in questions.iterrows():
    print(f"{row['Kod']}: {row['Pytanie']}")
```

```python
# P1_4 = "Nie czuję się samotny/a" -> deklaracja samotności
crosstab = pandas.crosstab(human_df["P3"], human_df["P1_4"])
percentage = crosstab.div(crosstab.sum(axis=1), axis=0) * 100
print(percentage.round(1))

percentage.plot(kind="bar", figsize=(12, 6))
plt.xlabel("Stan cywilny")
plt.ylabel("Odsetek osób [%]")
plt.title("Deklaracja samotności według stanu cywilnego")
plt.xticks(rotation=20, ha="right")
plt.legend(title="P1_4: Nie czuję się samotny/a")
plt.tight_layout()
plt.show()
```
